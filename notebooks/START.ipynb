## Рабочая папка

Создавайте ноутбуки в директории `/work` в JupyterLab, и они появятся в папке `jupyter-work` в корне проекта.

Это позволяет переносить материалы за пределы лаборатории и покрывать их системой контроля версий длы отслеживания прогресса.
